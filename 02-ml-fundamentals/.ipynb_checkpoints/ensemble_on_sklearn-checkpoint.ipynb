## Ensemble Learning On Sk Learn
- Bagging → train many models independently. Reduce variance.
- Boosting → train models sequentially. Fix previous mistakes. Reduce bias.
- Voting → combine predictions from different models.
- Stacking → use another model to learn how combine predictions.


..

### Loading Datasets And Preperations

````python
from sklearn.datasets import load_breast_cancer,fetch_california_housing
from sklearn.model_selection import train_test_split
````

````python
breast_cancer = load_breast_cancer()
````

````python
california_housing=fetch_california_housing()
````

````python
X=breast_cancer["data"]
y=breast_cancer["target"]
````

````python
X_housing=california_housing["data"]
y_housing=california_housing["target"]
````

````python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
````

````python
X_train_housing, X_test_housing, y_train_housing, y_test_housing = train_test_split(X_housing, y_housing, test_size=0.33, random_state=42)
````

### Gradient Boosting (Classification) On Sklearn

````python
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import accuracy_score
````

````python
model = GradientBoostingClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=3,
    random_state=42
)
````

````python
model.fit(X_train,y_train)
````

````python
y_pred = model.predict(X_test)
````

````python
accuracy_score(y_test,y_pred)
````

````python
model.score(X_test,y_test)
````

````python
0.9521276595744681*100
````

### Stacked Generalization (Stacking)-(Classifier)

````python
from sklearn.ensemble import StackingClassifier,RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
````

````python
estimators =[
    ("rf",RandomForestClassifier()),
    ("svm",SVC(probability=True)),
]
````

````python
model = StackingClassifier(
    estimators=estimators,
    final_estimator=LogisticRegression()
)
````

````python
model.fit(X_train,y_train)
````

````python
model.score(X_test,y_test)
````

### Stacking Regressor

````python
from sklearn.ensemble import StackingRegressor
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor

estimators = [
    ("rf", RandomForestRegressor())
]

model = StackingRegressor(
    estimators=estimators,
    final_estimator=LinearRegression()
)
model.fit(X_train_housing,y_train_housing)
````

````python
model.score(X_test_housing,y_test_housing)
````

````python
def adaptive_ada_boost(n_estimators,learning_rate):
    model = AdaBoostClassifier(
    n_estimators=n_estimators,
    learning_rate=learning_rate,
    random_state=42,
    )
    model.fit(X_train,y_train)
    score = model.score(X_test,y_test)
    print("Score :",score)
    
````

# Cheat Sheet
# Ensemble Methods Cheat Sheet

---

# Bagging vs Boosting

## Bagging

> Goal: **Reduce variance**

- Independent trees
- Parallel training
- Bootstrap sampling
- Less overfitting
- Stable models

**Examples**
- `RandomForestClassifier`
- `RandomForestRegressor`
- `BaggingClassifier`
- `BaggingRegressor`

---

## Boosting

> Goal: **Reduce bias**

- Sequential trees
- Each model fixes previous mistakes
- Weighted learning
- Higher predictive power
- More sensitive to hyperparameters

**Examples**
- `AdaBoostClassifier`
- `AdaBoostRegressor`
- `GradientBoostingClassifier`
- `GradientBoostingRegressor`
- `HistGradientBoostingClassifier`
- `HistGradientBoostingRegressor`

---

# Voting vs Stacking

## Voting

> Combine predictions directly

### Characteristics
- Simple
- Fast
- No meta-model
- Easy to train

**Models**
- `VotingClassifier`
- `VotingRegressor`

---

## Stacking

> Train meta-model on base model predictions

### Characteristics
- Learns best combination
- Usually higher accuracy
- Slower training
- More complex

**Models**
- `StackingClassifier`
- `StackingRegressor`

---

# Random Forest vs Gradient Boosting

## Random Forest

### Advantages
- Bagging
- Parallel training
- Less overfitting
- Fast training
- Strong default choice

**Model**
- `RandomForestClassifier`
- `RandomForestRegressor`

---

## Gradient Boosting

### Advantages
- Boosting
- Sequential training
- Usually higher accuracy
- Requires more tuning
- Slower training

**Models**
- `GradientBoostingClassifier`
- `GradientBoostingRegressor`
- `HistGradientBoostingClassifier`
- `HistGradientBoostingRegressor`

---

# What to Use When

## Small dataset

**Recommended**
- `GradientBoostingClassifier`
- `GradientBoostingRegressor`

---

## Large tabular dataset (>10,000 samples)

**Recommended**
- `HistGradientBoostingClassifier`
- `HistGradientBoostingRegressor`

---

## Need strong baseline

**Recommended**
- `RandomForestClassifier`
- `RandomForestRegressor`

---

## Need reduce variance

**Recommended**
- `BaggingClassifier`
- `BaggingRegressor`

---

## Already have several good models

**Recommended**
- `VotingClassifier`
- `VotingRegressor`

---

## Want highest possible accuracy

**Recommended**
- `StackingClassifier`
- `StackingRegressor`

---

## Need simple boosting

**Recommended**
- `AdaBoostClassifier`
- `AdaBoostRegressor`

---

# Quick Summary

| Situation | Best Choice |
|------------|-------------|
| Reduce variance | Bagging |
| Reduce bias | Boosting |
| Simple ensemble | Voting |
| Best ensemble accuracy | Stacking |
| Safe default | Random Forest |
| Highest accuracy on tabular data | Gradient Boosting / HistGradientBoosting |
| Large dataset | HistGradientBoosting |
| Small dataset | GradientBoosting |
